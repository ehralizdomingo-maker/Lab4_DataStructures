```python
# Exercise 1
LAST_NAME = "Domingo"
FAVORITE_ARTIST = "Taylor Swift"

# Generated Test Data
test_sources = ["T1", "T2", "T3", "T4", "T5"]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Generate numbers based on string lengths/positions
seed_num = len(LAST_NAME)
measurement_values = [
    seed_num * 10,
    seed_num * 15 + len(FAVORITE_ARTIST),
    85,
    seed_num * 20,
    95,
]

# Create raw records including a repeated entry
raw_records = [
    {"source": "T1", "category": "TEMPERATURE", "value": measurement_values[0]},
    {"source": "T2", "category": "VOLTAGE", "value": measurement_values[1]},
    {
        "source": "T1",
        "category": "TEMPERATURE",
        "value": measurement_values[0],
    },  # Repeated record
    {"source": "T3", "category": "CURRENT", "value": measurement_values[2]},
    {"source": "T4", "category": "PRESSURE", "value": measurement_values[3]},
    {"source": "T5", "category": "SPEED", "value": measurement_values[4]},
]

print("=== GENERATED TEST DATA ===")
print(raw_records)

# Organized Test Data
organized_data = {}
for record in raw_records:
  src = record["source"]
  if src not in organized_data:
    organized_data[src] = []
  organized_data[src].append(record)

print("\n=== ORGANIZED TEST DATA ===")
print(organized_data)

# Repeated/Distinct Data
unique_records = []
duplicate_records = []
seen = set()

for record in raw_records:
  identifier = (record["source"], record["category"], record["value"])
  if identifier in seen:
    duplicate_records.append(record)
  else:
    seen.add(identifier)
    unique_records.append(record)

print("\n=== DISTINCT/REPEATED RECORDS ===")
print("Distinct Records:", unique_records)
print("Repeated Records:", duplicate_records)

# Numerical and Categorical Summaries
all_values = [r["value"] for r in unique_records]
total_sum = sum(all_values)
average_value = total_sum / len(all_values)

print("\n=== NUMERICAL SUMMARY AND CATEGORICAL SUMMARY ===")
print(f"Sum: {total_sum} | Average: {average_value:.2f}")

# Processed Results
processed_results = unique_records
print("\n=== PROCESSED RESULTS ===")
print(processed_results)

# Ordered Results
ordered_results = sorted(processed_results, key=lambda x: x["value"])
print("\n=== ORDERED RESULTS ===")
print(ordered_results)

# Final Summary
print("\n=== FINAL SUMMARY ===")
print(f"Total Distinct Records: {len(unique_records)}")
print(f"Total Repeated Records: {len(duplicate_records)}")
```

```python
# EXERCISE 2

LAST_NAME = "Domingo"
FAVORITE_ARTIST = "Taylor Swift"

# Generated Performance Records
seed_num = len(LAST_NAME)
scores_list = [
    seed_num * 10 + ord(FAVORITE_ARTIST[0]),
    seed_num * 10 + ord(FAVORITE_ARTIST[1]),
    seed_num * 10 + ord(FAVORITE_ARTIST[2]),
    seed_num * 10 + ord(FAVORITE_ARTIST[3]),
    seed_num * 10 + ord(FAVORITE_ARTIST[4]),
]

student_names = [
    f"{LAST_NAME}_Prog",
    f"{LAST_NAME}_Math",
    f"{LAST_NAME}_Elec",
    f"{LAST_NAME}_Phys",
    f"{LAST_NAME}_Eng",
]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

performance_records = [
    {"name": student_names[0], "subject": subjects[0], "score": scores_list[0]},
    {"name": student_names[1], "subject": subjects[1], "score": scores_list[1]},
    {
        "name": student_names[0],
        "subject": subjects[0],
        "score": scores_list[0],
    },  # Repeated record
    {"name": student_names[2], "subject": subjects[2], "score": scores_list[2]},
    {"name": student_names[3], "subject": subjects[3], "score": scores_list[3]},
    {"name": student_names[4], "subject": subjects[4]},  # Incomplete record (missing score)
]

print("=== GENERATED PERFORMANCE RECORDS ===")
print(performance_records)

# Organized Student/Subject Data
valid_records = []
incomplete_records = []

for rec in performance_records:
  if "score" in rec:
    valid_records.append(rec)
  else:
    incomplete_records.append(rec)

print("\n=== ORGANIZED STUDENT/SUBJECT DATA ===")
print("Valid Records:", valid_records)
print("Incomplete Records:", incomplete_records)

# Repeated/Incomplete Records
seen_records = set()
unique_perf = []
repeated_perf = []

for rec in valid_records:
  identifier = (rec["name"], rec["subject"], rec["score"])
  if identifier in seen_records:
    repeated_perf.append(rec)
  else:
    seen_records.add(identifier)
    unique_perf.append(rec)

print("\n=== REPEATED/INCOMPLETE RECORDS ===")
print("Distinct Records:", unique_perf)
print("Repeated Records:", repeated_perf)
print("Incomplete Records:", incomplete_records)

# Performance Categories
categorized_performance = []
for rec in unique_perf:
  status = "PASSED" if rec["score"] >= 75 else "FAILED"
  categorized_performance.append({**rec, "category": status})

print("\n=== PERFORMANCE CATEGORIES ===")
print(categorized_performance)

# Student and Subject Summaries
scores_only = [r["score"] for r in unique_perf]
highest_score = max(scores_only)
lowest_score = min(scores_only)
average_score = sum(scores_only) / len(scores_only)

print("\n=== STUDENT AND SUBJECT SUMMARIES ===")
print(
    f"Highest Score: {highest_score} | Lowest Score: {lowest_score} | Average"
    f" Score: {average_score:.2f}"
)

# Ordered Results
ordered_performance = sorted(
    categorized_performance, key=lambda x: x["score"], reverse=True
)
print("\n=== ORDERED RESULTS ===")
print(ordered_performance)

# Final Output / Summary Report
print("\n=== FINAL STUDENT PERFORMANCE REPORT ===")
print(
    f"Processed {len(unique_perf)} unique records, {len(repeated_perf)}"
    f" repeated records, and {len(incomplete_records)} incomplete records"
    " successfully."
)
```

```python
# EXERCISE 3

LAST_NAME = "Domingo"
FAVORITE_ARTIST = "Taylor Swift"

seed_num = len(LAST_NAME)
quantities = [
    seed_num + (ord(FAVORITE_ARTIST[0]) % 5),
    seed_num + (ord(FAVORITE_ARTIST[1]) % 5),
    seed_num + (ord(FAVORITE_ARTIST[2]) % 5),
    seed_num + (ord(FAVORITE_ARTIST[3]) % 5),
    seed_num + (ord(FAVORITE_ARTIST[4]) % 5),
]

customers = [
    f"{LAST_NAME}_C1",
    f"{LAST_NAME}_C2",
    f"{LAST_NAME}_C3",
    f"{LAST_NAME}_C4",
    f"{LAST_NAME}_C5",
]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = [50.0, 20.0, 300.0, 1200.0, 450.0]

# Generated Transaction Records 
transactions = [
    {
        "customer": customers[0],
        "product": products[0],
        "qty": quantities[0],
        "price": prices[0],
    },
    {
        "customer": customers[1],
        "product": products[1],
        "qty": quantities[1],
        "price": prices[1],
    },
    {
        "customer": customers[0],
        "product": products[0],
        "qty": quantities[0],
        "price": prices[0],
    },  # Repeated purchase
    {
        "customer": customers[2],
        "product": products[2],
        "qty": quantities[2],
        "price": prices[2],
    },
    {
        "customer": customers[3],
        "product": products[3],
        "qty": quantities[3],
        "price": prices[3],
    },
]

print("=== GENERATED TRANSACTION RECORDS ===")
print(transactions)

# 2. Organized Transaction Data
organized_transactions = []
for t in transactions:
  total_val = t["qty"] * t["price"]
  organized_transactions.append(
      {
          "customer": t["customer"],
          "product": t["product"],
          "qty": t["qty"],
          "price": t["price"],
          "total_value": total_val,
      }
  )

print("\n=== ORGANIZED TRANSACTION DATA ===")
print(organized_transactions)

# Repeated/Distinct Transactions
unique_transactions = []
repeated_purchases = []
trans_seen = set()

for t in organized_transactions:
  key = (t["customer"], t["product"], t["qty"])
  if key in trans_seen:
    repeated_purchases.append(t)
  else:
    trans_seen.add(key)
    unique_transactions.append(t)

print("\n=== REPEATED/DISTINCT TRANSACTIONS ===")
print("Distinct Transactions:", unique_transactions)
print("Repeated Purchases:", repeated_purchases)

# Transaction Values
print("\n=== TRANSACTION VALUES ===")
for i, t in enumerate(unique_transactions, 1):
  print(
      f"Transaction {i} ({t['customer']} - {t['product']}): Qty:"
      f" {t['qty']} | Price: ₱{t['price']} | Total Value: ₱{t['total_value']:.2f}"
  )

# Customer Summaries
customer_summary = {}
for t in unique_transactions:
  cust = t["customer"]
  customer_summary[cust] = customer_summary.get(cust, 0) + t["total_value"]

print("\n=== CUSTOMER SUMMARIES ===")
print(customer_summary)

# Product Summaries
product_summary = {}
for t in unique_transactions:
  prod = t["product"]
  product_summary[prod] = product_summary.get(prod, 0) + t["total_value"]

print("\n=== PRODUCT SUMMARIES ===")
print(product_summary)

# Ordered Results (Sorted by total value descending)
ordered_transactions = sorted(
    unique_transactions, key=lambda x: x["total_value"], reverse=True
)
print("\n=== ORDERED RESULTS ===")
print(ordered_transactions)

# Overall Transaction Summary
grand_total = sum(t["total_value"] for t in unique_transactions)
print("\n=== OVERALL TRANSACTION SUMMARY ===")
print(
    f"Successfully processed {len(unique_transactions)} unique transactions"
    f" and {len(repeated_purchases)} repeated purchases. Grand Total Revenue:"
    f" ₱{grand_total:.2f}"
)
```
